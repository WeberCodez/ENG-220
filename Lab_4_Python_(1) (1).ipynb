# Data Libraries in Python

Numpy Basics

```python
# Arrays vs Lists
nums_list = [1,2,3,4,5]
nums_array = np.array([1,2,3,4,5])

print("List * 2:", nums_list * 2)     # repetition
print("Array * 2:", nums_array * 2)   # elementwise multiply
```

```python
# Array creation
a = np.zeros((2,3))
b = np.ones((2,3))
c = np.eye(3)    # identity matrix
d = np.arange(0,10,2)
e = np.linspace(0,1,5)

print("zeros:\n", a)
print("\nones:\n", b)
print("\nidentity:\n", c)
print("\narange:", d)
print("\nlinspace:", e)
```

```python
# Operations
x = np.array([10,20,30,40])
y = np.array([1,2,3,4])

print("Add:", x + y)
print("Multiply:", x * y)
print("Dot product:", x @ y)
print("Mean:", x.mean())
print("Reshape:", np.arange(12).reshape(3,4))
```

Pandas Basics

```python
# Create DataFrame
data = {
    "Name":["Ava","Liam","Noah","Maya"],
    "Age":[22,23,21,24],
    "Major":["CS","Math","EE","CS"]
}
df = pd.DataFrame(data)
print(df)

# Inspect
print(df.head())
df.info()               # .info() prints on its own; wrapping in print() adds an extra 'None'
print(df.describe(include="all"))

# Select
print(df["Name"])
print(df[["Name","Age"]])

# Filter
print(df[df["Age"] > 22])

# Group
print(df.groupby("Major")["Age"].mean())
```

```python
# Load real dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
iris = pd.read_csv(url)
iris.head()
```

```python
#Indexing and selecting data

import pandas as pd
import seaborn as sns

# NOTE: this intentionally OVERWRITES the 'iris' loaded earlier from the CSV URL.
# We reload it here via seaborn's built-in dataset for convenience going forward.
iris = sns.load_dataset("iris")

# Using .loc and .iloc
print(iris.loc[0, "species"])   # label-based
print(iris.iloc[0, 2])          # position-based
```

```python
# Sorting
# Sort by sepal_length
iris_sorted = iris.sort_values(by="sepal_length", ascending=False)
print(iris_sorted.head())
```

```python
# Adding and dropping columns

iris["sepal_area"] = iris["sepal_length"] * iris["sepal_width"]
print(iris.head())

print("\n")

iris = iris.drop(columns=["sepal_area"])
print(iris.head())
```

Matplotlib Basics

```python
# Line plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Line Plot Example")
plt.legend()
plt.show()
```

```python
# Bar plot
majors = df["Major"].value_counts()
majors.plot(kind="bar", color="skyblue")
plt.title("Count by Major")
plt.show()
```

```python
# Histogram
iris["sepal_length"].plot(kind="hist", bins=20, color="orange")
plt.title("Sepal Length Distribution")
plt.show()
```

```python
# Scatter
plt.scatter(iris["sepal_length"], iris["petal_length"],
            c=iris["sepal_width"], cmap="viridis", alpha=0.7)
plt.colorbar(label="Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Colored Scatter Plot")
plt.show()
```

```python
# Pie Chart
counts = iris["species"].value_counts()
plt.pie(counts, labels=counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Species Distribution")
plt.show()
```

Seaborn Basics

```python
# Scatterplot
sns.scatterplot(data=iris, x="sepal_length", y="petal_length", hue="species")
plt.title("Iris Scatterplot")
plt.show()
```

```python
# Histogram/Distribution
sns.histplot(iris["sepal_length"], kde=True, bins=20, color="green")
plt.title("Sepal Length Histogram with KDE")
plt.show()
```

```python
# Boxplot
sns.boxplot(data=iris, x="species", y="sepal_length")
plt.title("Sepal Length by Species")
plt.show()
```

File Handling

```python
# Write text to a file
with open("students.txt", "w") as f:
    f.write("Name,Age,Major\n")
    f.write("Ava,22,CS\n")
    f.write("Liam,23,Math\n")

print("File written.")
```

```python
# Reading from a file
with open("students.txt", "r") as f:
    content = f.read()
print(content)
```

```python
# Reading line by line
with open("students.txt", "r") as f:
    for line in f:
        print("Line:", line.strip())
```
